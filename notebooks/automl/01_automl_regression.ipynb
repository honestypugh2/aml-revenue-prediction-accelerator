# Azure Machine Learning Automated ML (regression)

Build and submit an AutoML **regression** job to predict
`actual_month_end_net_revenue` using the Azure ML Python SDK v2.

**All data is synthetic.** This notebook requires the `azure` extra and a
configured workspace to actually submit; the spec-building cells run offline.

```python
from revenue_prediction.config.loader import load_settings
from revenue_prediction.automl import build_automl_job_spec

settings = load_settings('dev')
spec = build_automl_job_spec(
    settings.automl,
    settings.azure_ml,
    training_data_asset='azureml:revenue_snapshots_mltable@latest',
)
spec
```

## Prepare an MLTable (offline)

AutoML consumes an MLTable. Generate the dataset and write an `MLTable` file
alongside the data, then register it as a data asset.

```python
from pathlib import Path
from revenue_prediction.data.io import materialise_default_datasets

paths = materialise_default_datasets(settings.data)
mltable_dir = Path('data/synthetic')
mltable_yaml = '''type: mltable
paths:
  - file: ./revenue_snapshots.parquet
transformations:
  - read_parquet
'''
(mltable_dir / 'MLTable').write_text(mltable_yaml)
print('Wrote', mltable_dir / 'MLTable')
```
